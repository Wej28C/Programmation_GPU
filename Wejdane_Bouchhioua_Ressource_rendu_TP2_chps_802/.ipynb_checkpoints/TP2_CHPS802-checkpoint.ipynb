**Version 1** :

**Version 2**  :
L'accès non coalisé à la mémoire entraîne un plus grand nombre de transactions en mémoire, ce qui augmente le temps d'exécution en raison de l'utilisation inefficace de la bande passante de la mémoire.

**Interprétations :**


*  **Version 1**:
Le schéma d'accès à la mémoire est optimal en raison de la coalescence, ce qui entraîne une latence minimale de la mémoire.
le thread 0 lit d_z[0], le thread 1 lit d_z[1], ..., le thread 31 lit d_z[31].
* **Version 2** :
Les threads lisent des éléments non contigus en raison de l'intervalle de 1.
Le thread 0 lit d_z[0], le thread 1 lit d_z[2*N], le thread 2 lit d_z[4*N], ..., ce qui entraîne un accès inefficace à la mémoire.



**Question5:** the effective transfer rate in GB/s

Avec N est le nombre de pas de temps :
<br>
* Mémoire totale accédée par thread =
2 * N * sizeof(float) bytes
* Mémoire totale accédée par bloc = le nombre de threads par bloc * 2 * N * sizeof(float) bytes
* Mémoire totale accédée pour tous les blocs = nombre de blocs * le nombre de threads par bloc * 2 * N * sizeof(float) bytes
<br>
* nombre de blocs: NPATH/128 = 75 000
  * NPATH= 9600000
* nombre de threads par bloc: 128
* N= 100
  * Mémoire par thread = 2* 100 * 4=800
<br> => **Memoire totale** =
75000 *128 * 800 = 7 680 000 000= 7.68GB

<br>
** Transfer rate ** = Memoire Total / Kernel Execution time
= 7.68(GB) /0,03124666(s)= 245,786GB/s
<br> (Nous utilisons le temps d'execution de Version 1 car il est plus performant)
<br> Je travaille sur Google Collab, je ne connais pas quel GPU j'utilise je ne pourrais pas faire la comparaison pour la capacité maximale.
Mais si nous supposons que nous travaillons avec The V100 GPU qui est  capable de 900GB/s fourni dans "NVIDIA datasheet3 sur le sujet de TP, nous remarquons que notre Transfer rate est bien faible
